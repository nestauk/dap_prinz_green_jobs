```python
all_sectors_aggs = pd.DataFrame()
for i,g in welsh_mid_salary_green_and_not.groupby('sector'):
    w_v = g[g['country']=='Wales'].iloc[0]['value']
    e_v = g[g['country']=='England'].iloc[0]['value']
    diff = e_v - w_v
    all_sectors_aggs = pd.concat([all_sectors_aggs, pd.DataFrame(
        {'sector': [i],
         'diff':[diff],
         'Type': g.iloc[0]['Type'],
         'mean': g['value'].mean(),
         'england_value': e_v,
         'wales_value': w_v,
         'diff_prop_england': diff/e_v
         
        }
    )])
```
